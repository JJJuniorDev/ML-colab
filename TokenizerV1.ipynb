```python
import re
with open("the-verdict.txt", "r", encoding="utf-8") as f:
 raw_text=f.read()
preprocessed= re.findall(r"\w+|[^\w\s]", raw_text)
preprocessed= [item.strip() for item in preprocessed if item.strip()]
print(len(preprocessed))
print(preprocessed[:30])
```

```python
all_words=sorted(set(preprocessed))
vocab_size=len(all_words)
print(vocab_size)
```

```python
from pygments import token
vocab={token:integer for integer, token in enumerate(all_words)}
for i, item in enumerate(vocab.items()):
  print(item)
```

```python
class SimpleTokenizerV1:
  def __init__(self, vocab):
    self.stringa_in_intero=vocab
    self.intero_in_stringa={i:s for s,i in vocab.items()}

  def fase1_encode(self,text):
    preprocessed=re.findall(r"\w+|[^\w\s]", text)
    preprocessed= [item.strip() for item in preprocessed if item.strip()]
    preprocessed= [item if item in self.stringa_in_intero else "<|unk|>" for item in preprocessed]
    ids= [self.stringa_in_intero[s] for s in preprocessed]
    return ids

  def fase2_decode(self,ids):
    text=" ".join([self.intero_in_stringa[i] for i in ids])
    text=re.sub(r'\s+([,.?!"()\'])', r'\1', text)
    return text
```

```python
tokenizer= SimpleTokenizerV1(vocab)
text= """"It's the last he painted, you know, "Mrs. Gisburn said with pardonable pride you you you you. figsa fuji fatzu"""
ids= tokenizer.fase1_encode(text)
print(ids)
```

```python
print(tokenizer.fase2_decode(ids))
```
