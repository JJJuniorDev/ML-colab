```python
import tensorflow as tf
import numpy as np
X= tf.range(-100, 100, 4)
tf.experimental.numpy.experimental_enable_numpy_behavior()
X = X.reshape(-1, 1)
y=X+10
X_train = X[:40] #80%
y_train= y[:40]
X_test= X[40:] #last 10, 20%
y_test=y[40:]
```

```python
tf.random.set_seed(42)
model_1=tf.keras.Sequential([
    tf.keras.layers.Dense(1)
])

model_1.compile(loss=tf.keras.losses.mae,
                optimizer=tf.keras.optimizers.SGD(),
                metrics=["mae"])

model_1.fit(X_train, y_train, epochs=100)
```

```python
y_preds_1=model_1.predict(X_test)
y_preds_1
```

```python
import matplotlib.pyplot as plt

def plot_predictions(train_data=X_train, train_labels=y_train,
                     test_data=X_test, test_labels=y_test,
                     predictions=y_preds_1):
  """
  Plots training data, test data e paragona predizioni con vere label
  """
  plt.figure(figsize=(10,7))
  plt.scatter(train_data, train_labels, c="b", label="Training data")
  plt.scatter(test_data, test_labels, c="g", label="Test data")
  plt.scatter(test_data, predictions, c="r", label="Predictions")
  plt.legend();
```

```python
plot_predictions(X_train, y_train, X_test, y_test, y_preds_1)
```

```python
model_2=tf.keras.Sequential([
    tf.keras.layers.Dense(10),
    tf.keras.layers.Dense(1)
])

model_2.compile(loss=tf.keras.losses.mae,
                optimizer=tf.keras.optimizers.SGD(),
                metrics=["mse"])

model_2.fit(X_train, y_train, epochs=100)
```

```python
y_preds_2=model_2.predict(X_test)
plot_predictions(predictions=y_preds_2)
```

```python
model_3=tf.keras.Sequential([
    tf.keras.layers.Dense(10),
    tf.keras.layers.Dense(10),
    tf.keras.layers.Dense(1)
])

model_3.compile(loss=tf.keras.losses.mae,
                optimizer=tf.keras.optimizers.SGD(),
                metrics=["mae"])

model_3.fit(X_train, y_train, epochs=50)

```

```python
y_preds_3=model_3.predict(X_test)
plot_predictions(predictions=y_preds_3)
```

```python
import pandas as pd
model_results= [["model_1", model_1.get_metrics_result()],
                ["model_2", model_2.get_metrics_result()],
                 ["model_3", model_3.get_metrics_result()]]

df = pd.DataFrame([{"model": m[0], **m[1]} for m in model_results])
df
```
