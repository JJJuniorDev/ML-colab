```python
from sklearn.datasets import make_circles
#creiamo 1k
n_samples=1000
X, y= make_circles(n_samples, noise=0.03, random_state=42)
```

```python
 import pandas as pd
 circles=pd.DataFrame({"X0":X[:, 0], "X1":X[:, 1], "label":y})
 circles
```

```python

import matplotlib.pyplot as plt
plt.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.RdYlBu);
```

```python
import tensorflow as tf
tf.random.set_seed(42)
model_1=tf.keras.Sequential([
    tf.keras.layers.Dense(1, activation="relu"),
    tf.keras.layers.Dense(1, activation="softmax")
])

model_1.compile(loss=tf.keras.losses.BinaryCrossentropy(),
                optimizer=tf.keras.optimizers.SGD(),
                metrics=["accuracy"])

model_1.fit(X, y, epochs=100)
```

```python
model_1.evaluate(X, y)
```

```python
#Proviamo a migliorare il modello
#per dati non lineari la relu con qualche layer e neurone extra
model_2=tf.keras.Sequential([
    tf.keras.layers.Dense(100, activation="relu"),
    tf.keras.layers.Dense(100, activation="relu"),
    tf.keras.layers.Dense(1)
])

model_2.compile(loss=tf.keras.losses.BinaryCrossentropy(),
                optimizer=tf.keras.optimizers.SGD(),
                metrics=["accuracy"])

history=model_2.fit(X, y, epochs=100)
```

```python
model_2.evaluate(X, y)
```

```python
A=tf.cast(tf.range(-10, 10), tf.float32)
plt.plot(A);
#funzione della sigmoid, come calcola
def sigmoid(x):
  return 1/ (1+tf.exp(-x))
```

```python
sigmoid(A)
```

```python
#vediamo come trasforma i dati
plt.plot(sigmoid(A))
```

```python
#ricreiamo la relu
def relu(x):
  return tf.maximum(0, x)

relu(A)
```

```python
plt.plot(relu(A))
```

```python
#miglioriamo il classifier dividendo in test e training set
#sono 1k totale quindi 800 train e 200 test
X_train, y_train=X[:800], y[:800]
X_test, y_test=X[800:], y[800:] #da 800 in poi
X_train.shape, X_test.shape
```

```python
tf.random.set_seed(42)

model_3=tf.keras.Sequential([
    tf.keras.layers.Dense(4, activation="relu"),
    tf.keras.layers.Dense(4, activation="relu"),
    tf.keras.layers.Dense(1, activation="sigmoid")
])

model_3.compile(loss="binary_crossentropy",
                optimizer=tf.keras.optimizers.Adam(learning_rate=0.01),
                metrics=["accuracy"])

history=model_3.fit(X_train, y_train, epochs=25, verbose=0)

```

```python
y_pred=model_3.evaluate(X_test, y_test)
```

```python
#mostriamo le loss
pd.DataFrame(history.history).plot()
plt.title("Model loss curves")
```

```python
#troviamo il lr ideale usando callbacks
model_4=tf.keras.Sequential([
    tf.keras.layers.Dense(4, activation="relu"),
    tf.keras.layers.Dense(4, activation="relu"),
    tf.keras.layers.Dense(1, activation="sigmoid")
])

model_4.compile(loss=tf.keras.losses.BinaryCrossentropy,
                optimizer=tf.keras.optimizers.Adam(),
                metrics=["accuracy"])

#creiamo lr callback
lr_scheduler=tf.keras.callbacks.LearningRateScheduler(lambda epoch: 1e-4 * 10**(epoch/20))

history1=model_4.fit(X_train, y_train, epochs=100, callbacks=[lr_scheduler], verbose=0)
```

```python
pd.DataFrame(history1.history).plot(figsize=(10,7), xlabel="epochs")
```

```python
#plottiamo il lr versus loss
lrs= 1e-4 * (10 ** (tf.range(100)/20))
plt.figure(figsize=(10,7))
plt.semilogx(lrs, history1.history["loss"])
plt.xlabel("Learning Rate")
plt.ylabel("Loss")
plt.title("LR vs Loss")
```
